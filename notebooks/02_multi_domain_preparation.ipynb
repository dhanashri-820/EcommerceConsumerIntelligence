```python
#review collection from dataset
home_samples = []

for row in home_stream:
    home_samples.append(row)

    if len(home_samples) >= 15000:
        break

print("Reviews collected:", len(home_samples))
```

```python
#converting to dataframe for better visualization and analysis
home_df = pd.DataFrame(home_samples)

print("Shape:", home_df.shape)
```

```python
print(home_df["rating"].value_counts().sort_index())
```

```python
def rating_to_sentiment(rating):
    if rating <= 2:
        return "negative"
    elif rating == 3:
        return "neutral"
    else:
        return "positive"


home_df["sentiment"] = home_df["rating"].apply(rating_to_sentiment)

print(home_df["sentiment"].value_counts())
```

```python
#balancing of dataset so that it should not be biased towards any particular class. We will collect 5000 reviews for each sentiment class (negative, neutral, positive) to create a balanced dataset.
from collections import defaultdict

balanced_reviews = {
    "negative": [],
    "neutral": [],
    "positive": []
}

for row in home_stream:
    rating = row["rating"]

    if rating <= 2 and len(balanced_reviews["negative"]) < 5000:
        balanced_reviews["negative"].append(row)

    elif rating == 3 and len(balanced_reviews["neutral"]) < 5000:
        balanced_reviews["neutral"].append(row)

    elif rating >= 4 and len(balanced_reviews["positive"]) < 5000:
        balanced_reviews["positive"].append(row)

    # Stop once all three classes reach 5000
    if (
        len(balanced_reviews["negative"]) >= 5000
        and len(balanced_reviews["neutral"]) >= 5000
        and len(balanced_reviews["positive"]) >= 5000
    ):
        break

print("Negative:", len(balanced_reviews["negative"]))
print("Neutral:", len(balanced_reviews["neutral"]))
print("Positive:", len(balanced_reviews["positive"]))
```

```python
#combining the balanced reviews into a single list
home_balanced = (
    balanced_reviews["negative"]
    + balanced_reviews["neutral"]
    + balanced_reviews["positive"]
)

home_balanced_df = pd.DataFrame(home_balanced)

print("Final shape:", home_balanced_df.shape)
```

```python
#creating sentimnent column in the balanced dataframe
home_balanced_df["sentiment"] = home_balanced_df["rating"].apply(
    rating_to_sentiment
)
```

```python
print(home_balanced_df["sentiment"].value_counts())
```

```python
#shuffling the balance rows in datset so that the model should not be 
home_balanced_df = home_balanced_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
from huggingface_hub import hf_hub_download

appliances_file = hf_hub_download(
    repo_id="McAuley-Lab/Amazon-Reviews-2023",
    filename="raw_review_Appliances/full-00000-of-00002.parquet",
    repo_type="dataset",
    revision="e4458357e3499c762fb83a8c721fde557b7d0e8d"
)

print("Appliances file downloaded!")
print(appliances_file)
```

```python
import pyarrow.parquet as pq

appliances_parquet = pq.ParquetFile(appliances_file)

print("Rows:", appliances_parquet.metadata.num_rows)
print("Columns:", appliances_parquet.metadata.num_columns)
```

```python
#columns in the appliances parquet file
import pandas as pd

appliances_df = pd.read_parquet(appliances_file)

print("Shape:", appliances_df.shape)
print(appliances_df.columns.tolist())
```

```python
#rating
print(
    appliances_df["rating"]
    .value_counts()
    .sort_index()
)
```

```python
def rating_to_sentiment(rating):
    if rating <= 2:
        return "negative"
    elif rating == 3:
        return "neutral"
    else:
        return "positive"

appliances_df["sentiment"] = appliances_df["rating"].apply(
    rating_to_sentiment
)

print(appliances_df["sentiment"].value_counts())
```

```python
print("Negative:", (appliances_df["sentiment"] == "negative").sum())
print("Neutral:", (appliances_df["sentiment"] == "neutral").sum())
print("Positive:", (appliances_df["sentiment"] == "positive").sum())
```

```python
negative_df = appliances_df[
    appliances_df["sentiment"] == "negative"
].sample(n=5000, random_state=42)

neutral_df = appliances_df[
    appliances_df["sentiment"] == "neutral"
].sample(n=5000, random_state=42)

positive_df = appliances_df[
    appliances_df["sentiment"] == "positive"
].sample(n=5000, random_state=42)
```

```python
appliances_balanced_df = pd.concat(
    [negative_df, neutral_df, positive_df],
    ignore_index=True
)
```

```python
#shuffle
appliances_balanced_df = appliances_balanced_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
print("Shape:", appliances_balanced_df.shape)
print(appliances_balanced_df["sentiment"].value_counts())
```

```python
appliances_balanced_df.to_csv(
    "../data/processed/appliances_balanced.csv",
    index=False
)

print("Appliances balanced dataset saved!")
```

```python
#for sports and outdoors reviews
from huggingface_hub import list_repo_files

files = list_repo_files(
    repo_id="McAuley-Lab/Amazon-Reviews-2023",
    repo_type="dataset"
)

sports_files = [
    f for f in files
    if "Sports_and_Outdoors" in f
]

for f in sports_files[:50]:
    print(f)
```

```python
for f in sports_files:
    print(f)
```

```python
sports_url = (
    "https://huggingface.co/datasets/"
    "McAuley-Lab/Amazon-Reviews-2023/resolve/main/"
    "raw/review_categories/Sports_and_Outdoors.jsonl"
)

print(sports_url)
```

```python
import requests
import json

response = requests.get(
    sports_url,
    stream=True
)

print("Status code:", response.status_code)
```

```python
#collecting only balanced reviews for sports and outdoors dataset
sports_samples = {
    "negative": [],
    "neutral": [],
    "positive": []
}

for line in response.iter_lines():

    if not line:
        continue

    review = json.loads(line)

    rating = review.get("rating")

    if rating is None:
        continue

    if rating <= 2 and len(sports_samples["negative"]) < 5000:
        sports_samples["negative"].append(review)

    elif rating == 3 and len(sports_samples["neutral"]) < 5000:
        sports_samples["neutral"].append(review)

    elif rating >= 4 and len(sports_samples["positive"]) < 5000:
        sports_samples["positive"].append(review)

    if (
        len(sports_samples["negative"]) >= 5000
        and len(sports_samples["neutral"]) >= 5000
        and len(sports_samples["positive"]) >= 5000
    ):
        break

print("Negative:", len(sports_samples["negative"]))
print("Neutral:", len(sports_samples["neutral"]))
print("Positive:", len(sports_samples["positive"]))
```

```python
#creating a DataFrame for the balanced sports and outdoors reviews
sports_balanced = (
    sports_samples["negative"]
    + sports_samples["neutral"]
    + sports_samples["positive"]
)

sports_df = pd.DataFrame(sports_balanced)

print("Shape:", sports_df.shape)
```

```python
#adding sentiment column to the sports and outdoors balanced DataFrame
sports_df["sentiment"] = sports_df["rating"].apply(
    rating_to_sentiment
)
```

```python
print(sports_df["sentiment"].value_counts())
```

```python
#shuffling the sports and outdoors balanced DataFrame
sports_df = sports_df.sample(
    frac=1,
    random_state=42
).reset_index(drop=True)
```

```python
sports_df.to_csv(
    "../data/processed/sports_outdoors_balanced.csv",
    index=False
)

print("Sports & Outdoors balanced dataset saved!")
```
